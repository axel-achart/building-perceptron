## **2. Draw me an artificial neuron's monitoring**

##### 1. **Qu’est ce qu’un Perceptron ? Quel est le lien entre un neurone biologique et un perceptron ?** </br>
Un perceptron est un algorithme d'apprentissage supervisé composé d'un seul neurone et utilisé en intelligence artificielle et en apprentissage automatique pour la classification binaire. </br>
Le perceptron est une simplification informatique du neurone biologique qui reçoit des signaux (dendrites), les pondère selon leur importance (synapses) et les additionne (corps cellulaire) pour émettre ou non un signal de sortie (axone) selon le principe du « tout ou rien ».


##### 2. **Quelle est la fonction mathématique du Perceptron et son usage ?** </br>

**Définition des termes**

| Symbole | Nom | Rôle |
|:---:|---|---|
| $x = (x_1, x_2, \dots, x_n)$ | **Vecteur d'entrées** | Les données fournies au neurone (une valeur par caractéristique) |
| $w = (w_1, w_2, \dots, w_n)$ | **Poids synaptiques** | L'importance accordée à chaque entrée $x_i$ |
| $b$ (ou $-\theta$) | **Biais** | Décale la frontière de décision (seuil d'activation) |
| $z$ | **Somme pondérée** | Le résultat du calcul avant activation |
| $y$ | **Sortie** | La classe prédite : $1$ ou $0$ |

**Fonction mathématique (somme pondérée)**

$$
z \;=\; \sum_{i=1}^{n} w_i \, x_i \;+\; b \;=\; w \cdot x + b
$$

> On multiplie chaque entrée par son poids, on additionne le tout, puis on ajoute le biais.


##### 3. **Donnez une ou plusieurs règles d’apprentissage du Perceptron.** </br>
- Les données doivent être linéairement séparables.
- Les poids et le biais ne sont modifiés que si le perceptron commet une erreur de prédiction.
- L'apprentissage s'arrête (converge) uniquement lorsque toutes les données du jeu d'entraînement sont parfaitement classées, ce qui nécessite la première règle.


##### 4. **Le perceptron utilise généralement une fonction d’activation, laquelle ?** </br>
**Fonction d'activation (seuil de Heaviside)**

$$
y \;=\; F(z) \;=\;
\begin{cases}
1 & \text{si } z \geq 0 \\[4pt]
0 & \text{si } z < 0
\end{cases}
$$

> Si la somme pondérée atteint le seuil, le neurone « s'active » et renvoie $1$ ; sinon il renvoie $0$ (ou $-1$ selon les conventions).


##### 5. **Quel est le processus d'entraînement du Perceptron ?** </br>
- L'initialisation
- Calcul de la sommation (y) et de l'activation (y^)
- Si y^ = t : bonne prédiction, on change rien et ça passe au suivant
- Sinon mettre à jour le poids avec la règle d'apprentissage
- Faire ce process pour toutes les données, et le passage entir des donnée s'appelle une époque
- Stopper le cycle d'apprentissage soit si il est passé par la totalité des données ou après x époques


##### 6. **Quelles sont les limites du Perceptron ?** </br>
- Incapacité à résoudre des problèmes qui ne sont pas linéairement séparables
- Incapacité à deviner plus, qu'une solution type "oui" ou "non" (il n'est qu'en binaire)
- S'adapte mal à de nouveaux types d'informations car il reste sur des données étiquetées, classification binaire seulement.
