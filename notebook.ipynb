### **What are them ?**
**Machine Learning =** It is a subset of artificial intelligence that allows computers to learn from data and improve their performance over time without being explicitly programmed with hard-coded rules.
</br>
**Deep Learning =** This is a subcategory of artificial intelligence and machine learning that uses multilayer artificial neural networks inspired by the workings of the human brain to analyze data and autonomously learn about rather complex subjects.
</br>

### **When do we need to use them ?**
**Machine Learning =** for analyzing well organized tables of numbers and data (such as Excel files).
</br>
**Deep Learning =** to analyze complex data such as images, videos, or text (like photographs or human conversations).
</br>

### **3 Different applications of the Deep Learning**
- NLP : Deep Learning is used for NLP to give the capacity for the machine, to know how to learn by themselves.
- Speech-to-text transcription (on smartphones, smart speakers) and text-to-speech synthesis.
- Analysis of medical images (X-rays, MRIs, CT scans) to detect tumors or early-stage abnormalities, as detailed by the [Brain Institute](https://institutducerveau.org/lexique/deep-learning-apprentissage-profond).
