## 2. Dessine-moi un neurone artificiel

### 1. Qu'est-ce qu'un Perceptron ?
Le Perceptron est un algorithme de classification binaire inspiré du fonctionnement simplifié d'un neurone biologique. Les entrées correspondent aux signaux reçus, les poids représentent leur importance, puis une fonction d'activation produit la sortie.

### 2. Fonction mathématique
Le Perceptron calcule : **z = Σ(wᵢxᵢ) + b**.

- x : entrées
- w : poids
- b : biais
- z : somme pondérée

### 3. Règle d'apprentissage
Les poids sont corrigés avec l'erreur : **w = w + learning_rate × erreur × x**.

### 4. Fonction d'activation
Le Perceptron utilise généralement une fonction seuil : 1 si le résultat est supérieur ou égal à 0, sinon 0.

### 5. Entraînement
Le modèle fait une prédiction, la compare à la vraie réponse puis modifie les poids et le biais si la prédiction est fausse. Le processus est répété plusieurs fois sur les données.

### 6. Limites
Un Perceptron simple ne peut résoudre correctement que des problèmes linéairement séparables. Il ne peut par exemple pas résoudre seul le problème XOR.

## 3. Test du Perceptron avec des données aléatoires

```python
import numpy as np
from perceptron import Perceptron

np.random.seed(42)
X = np.random.uniform(-2, 2, (100, 2))
y = (X[:, 0] + X[:, 1] > 0).astype(int)

model = Perceptron(learning_rate=0.1, epochs=100)
model.fit(X, y)
predictions = model.predict(X)

accuracy = np.mean(predictions == y)
print('Accuracy :', accuracy)
```

Le test utilise volontairement des données simples et linéairement séparables. Cela permet de vérifier que notre implémentation du Perceptron fonctionne avant de l'utiliser sur le dataset Breast Cancer Wisconsin.
