## **2. Draw me an artificial neuron's monitoring**

##### 1. **What is a Perceptron? What is the link between a biological neuron and a perceptron?** </br>
A perceptron is a supervised learning algorithm made of a single neuron, used in artificial intelligence and machine learning for binary classification. </br>
The perceptron is a computational simplification of the biological neuron: it receives signals (dendrites), weights them according to their importance (synapses) and sums them up (cell body) to emit or not an output signal (axon), following the "all-or-nothing" principle.


##### 2. **What is the Perceptron's mathematical function and what is it used for?** </br>

**Definition of terms**

| Symbol | Name | Role |
|:---:|---|---|
| $x = (x_1, x_2, \dots, x_n)$ | **Input vector** | The data fed to the neuron (one value per feature) |
| $w = (w_1, w_2, \dots, w_n)$ | **Synaptic weights** | The importance given to each input $x_i$ |
| $b$ (or $-\theta$) | **Bias** | Shifts the decision boundary (activation threshold) |
| $z$ | **Weighted sum** | The result of the computation before activation |
| $y$ | **Output** | The predicted class: $1$ or $0$ |

**Mathematical function (weighted sum)**

$$
z \;=\; \sum_{i=1}^{n} w_i \, x_i \;+\; b \;=\; w \cdot x + b
$$

> Each input is multiplied by its weight, everything is summed up, then the bias is added.


##### 3. **Give one or more learning rules of the Perceptron.** </br>
- The data must be linearly separable.
- The weights and the bias are only updated when the perceptron makes a prediction error.
- Learning stops (converges) only once every sample of the training set is perfectly classified, which requires the first rule.


##### 4. **The perceptron usually uses an activation function, which one?** </br>
**Activation function (Heaviside step function)**

$$
y \;=\; F(z) \;=\;
\begin{cases}
1 & \text{if } z \geq 0 \\[4pt]
0 & \text{if } z < 0
\end{cases}
$$

> If the weighted sum reaches the threshold, the neuron "fires" and returns $1$; otherwise it returns $0$ (or $-1$ depending on the convention).


##### 5. **What is the Perceptron's training process?** </br>
- Initialization
- Compute the weighted sum (y) and the activation (y^)
- If y^ = t: correct prediction, nothing changes and we move on to the next sample
- Otherwise, update the weights with the learning rule
- Repeat this process for all the data; one full pass over the data is called an epoch
- Stop the learning cycle either once it has gone through all the data or after x epochs


##### 6. **What are the limits of the Perceptron?** </br>
- Unable to solve problems that are not linearly separable
- Unable to predict more than a "yes" or "no" type of answer (it is binary only)
- Adapts poorly to new types of information since it relies on labeled data, binary classification only.
